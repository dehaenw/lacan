# Median molecules
LACAN uses a straightforward method to generate "median molecules": these are molecules resulting from the recombination of substructures from two compounds. The idea is that these molecules to some extent will have properties inbetween those of the two parents. More meaningfully, this can be used as a genetic operator in evolutionary strategies for molecule optimization

## crossing Fluoxetine with Sertraline

```python
from rdkit import Chem
from lacan import lacan
from lacan import breed
from rdkit.Chem import Draw
from rdkit.Chem.Draw import IPythonConsole
IPythonConsole.drawOptions.drawMolsSameScale=False

# load parent molecules
m1 = Chem.MolFromSmiles("c1cc(ccc1[C@@H]2CCNC[C@H]2COc3ccc4c(c3)OCO4)F")
m2 = Chem.MolFromSmiles("CNCCC(C1=CC=CC=C1)OC2=CC=C(C=C2)C(F)(F)F")

# load lacan profile
p = lacan.load_profile("chembl")
# generate median molecules
median_molecules = breed.breed(m1,m2,p,nmols=9,cuts=2)
# display the molecules
d = Draw.MolsToGridImage(median_molecules,molsPerRow=3)
display(d)
```

## Crossing atorvastatin and captopril


```python


# load parent molecules
m1 = Chem.MolFromSmiles("CC(C)C1=C(C(=C(N1CC[C@H](C[C@H](CC(=O)O)O)O)C2=CC=C(C=C2)F)C3=CC=CC=C3)C(=O)NC4=CC=CC=C4")
m2 = Chem.MolFromSmiles("O=C(O)[C@H]1N(C(=O)[C@H](C)CS)CCC1")

# load lacan profile
median_molecules = breed.breed(m1,m2,p,nmols=9,hacrange=(0.5,2.0),interprange=(0.1,0.9))
# display the molecules
d = Draw.MolsToGridImage(median_molecules,molsPerRow=3)
display(d)
```
